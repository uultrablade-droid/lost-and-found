## 💻 Example 1: Grade Calculator
Complete working example with functions.


```python
def calculate_grade(score):
    """Calculate letter grade from numeric score"""
    if score >= 80:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"

def calculate_average(scores):
    """Calculate average of a list of scores"""
    return sum(scores) / len(scores)

def print_grade_report(name, scores):
    """Print a complete grade report"""
    print(f"\n{'='*40}")
    print(f"Grade Report for {name}")
    print(f"{'='*40}")
    
    for i, score in enumerate(scores, 1):
        grade = calculate_grade(score)
        print(f"Subject {i}: {score:>5.1f} ({grade})")
    
    average = calculate_average(scores)
    final_grade = calculate_grade(average)
    print(f"{'-'*40}")
    print(f"Average:  {average:>5.1f} ({final_grade})")
    print(f"{'='*40}")

# Use the functions
student_scores = [85, 90, 78, 92, 88]
print_grade_report("Alice", student_scores)
```

## 💻 Example 2: Todo List Manager
Working example of a task management system.


```python
def add_task(todo_list, task):
    """Add a task to the todo list"""
    todo_list.append(task)
    print(f"✅ Added: {task}")

def remove_task(todo_list, task):
    """Remove a task from the todo list"""
    if task in todo_list:
        todo_list.remove(task)
        print(f"✅ Removed: {task}")
    else:
        print(f"❌ Task not found: {task}")

def show_tasks(todo_list):
    """Display all tasks"""
    print("\n📋 Todo List:")
    if not todo_list:
        print("  (empty)")
    else:
        for i, task in enumerate(todo_list, 1):
            print(f"  {i}. {task}")

def mark_complete(todo_list, task_number):
    """Mark a task as complete by moving it to a completed list"""
    if 1 <= task_number <= len(todo_list):
        task = todo_list.pop(task_number - 1)
        print(f"✅ Completed: {task}")
        return task
    else:
        print("❌ Invalid task number")
        return None

# Demonstrate usage
tasks = []
add_task(tasks, "Study Python")
add_task(tasks, "Buy groceries")
add_task(tasks, "Exercise")
show_tasks(tasks)

print("\n--- After completing task 1 ---")
mark_complete(tasks, 1)
show_tasks(tasks)
```
